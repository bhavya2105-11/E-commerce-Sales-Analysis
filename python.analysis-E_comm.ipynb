```python
#importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
#Loading the data
customers = pd.read_csv("customers.csv")
orders = pd.read_csv("orders.csv")
payments = pd.read_csv("payments.csv")
products = pd.read_csv("products.csv")
```

### Reading the data

```python
print(customers.head())
print(customers.shape)
```

```python
print(orders.head())
print(orders.shape)
```

```python
print(payments.head())
print(payments.shape)
```

```python
print(products.head())
print(products.shape)
```

### Checking Null Values

```python
print(customers.isnull().sum())
print(orders.isnull().sum())
print(payments.isnull().sum())  
print(products.isnull().sum())
```

### Removing Duplicates

```python
print(customers.duplicated().sum())
print(orders.duplicated().sum())
print(payments.duplicated().sum())
print(products.duplicated().sum())
```

```python
print("Before dropping duplicates:", orders.shape)
orders.drop_duplicates(inplace=True)#inplace=True will modify the original DataFrame instead of creating a new one
print("After dropping duplicates:", orders.shape)
```

### Removing Nulls ###

```python
orders = orders.dropna(subset = ['OrderDate' , 'Quantity']) #dropping rows with missing OrderDate and Quantity values
print((orders["Quantity"] <= 0).sum())
orders = orders[orders["Quantity"] > 0]
print("After dropping rows with missing OrderDate and Quantity values:", orders.shape)
print(orders.isnull().sum()) #checking for missing values again
```

### Replacing Dupilicates

```python
orders = orders.fillna({'Discount': 0}) #filling missing Discount values with 0
orders = orders.fillna({'PaymentMethod': 'Unknown'}) #filling missing PaymentMethod values with 'Unknown'
print("After filling missing Discount and PaymentMethod values:", orders.isnull().sum()) #checking for missing values again
```

```python
customers = customers.fillna({'City': 'Unknown'}) #filling missing City values with 'Unknown'
customers = customers.fillna({'Age': customers['Age'].median()}) #filling missing Age values with the median age
print("After filling missing City and Age values:", customers.isnull().sum()) #checking for missing values again
```

### Changing the Data type

```python
customers['SignupDate'] = pd.to_datetime(customers['SignupDate']) #converting SignupDate to datetime format
orders['OrderDate'] = pd.to_datetime(orders['OrderDate']) #converting OrderDate to datetime format
payments['PaymentDate'] = pd.to_datetime(payments['PaymentDate']) #converting PaymentDate to datetime format
```

### Validating the Data types

```python
print(customers.dtypes)
print(orders.dtypes)
print(payments.dtypes)
print(products.dtypes)
```

### Merging the files

```python
Df = pd.merge(orders, customers, on='CustomerID', how='left') #merging orders and customers dataframes on CustomerID
print(Df.head())
```

```python
Df = pd.merge(Df , payments, on='OrderID', how='left') #merging the previous dataframe with payments dataframe on OrderID
print(Df.head())
```

```python
Df = pd.merge(Df, products, on='ProductID', how='left') #merging the previous dataframe with products dataframe on productID
print(Df.head(10))
```

### Checking Null Again

```python
print(Df.isnull().sum()) #checking for missing values in the final merged dataframe
```

```python
missing = Df[Df['City'].isnull()] #filtering rows where City is missing
print(missing['CustomerID'].unique())
```

```python
Df = Df.fillna({'City': 'Unknown'}) #filling missing City values with 'Unknown'
Df = Df.fillna({'CustomerSegment': 'Unknown'}) #filling missing CustomerSegment values with 'Unknown'
print("After filling missing City and CustomerSegment values:", Df.isnull().sum()) #checking for missing values again
```

### Calculating The Revenue ###

```python
Df['Revenue'] = Df['Quantity'] * Df['UnitPrice'] * (1 - Df['Discount']) #calculating revenue for each order
print(Df[['OrderID', 'Quantity', 'UnitPrice', 'Discount', 'Revenue']].head(10)) #displaying the first 5 rows of the relevant columns
```

```python
print(Df["Status"].value_counts()) #counting the number of orders in each status category
print(pd.crosstab(Df['Status'], Df['PaymentMethod'])) #creating a cross-tabulation of Status and PaymentMethod
print(pd.crosstab(Df['Status'], Df['PaymentStatus'])) #creating a cross-tabulation of Status and PaymentStatus with row and column totals
```

```python
NewRevenue = Df[(Df['Status'] == 'Completed') & (Df['PaymentStatus'] == 'Paid')]
NewRevenue["City"] = NewRevenue["City"].replace("Mashad", "Mashhad")
print(NewRevenue["City"].value_counts())
print(NewRevenue['Revenue'].sum()) #calculating the total revenue from completed and paid orders
print(NewRevenue.shape) #calculating the number of completed and paid orders
```

### KPI Cards ###


```python
TotalRevenue = NewRevenue['Revenue'].sum() #calculating the total revenue from all orders
print("Total Revenue:", TotalRevenue)
TotalOrders = NewRevenue.shape[0] #calculating the total number of orders
print("Total Orders:", TotalOrders)
AverageRevenuePerOrder = TotalRevenue / TotalOrders #calculating the average revenue per order
print("Average Revenue Per Order:", AverageRevenuePerOrder)
TotalSoldProducts = NewRevenue['Quantity'].sum() #calculating the total number of products sold
print("Total Sold Products:", TotalSoldProducts)
AllOrders = Df.shape[0] #calculating the total number of orders in the dataset  
CancellationRate = Df[Df['Status'] == 'Cancelled'].shape[0] / AllOrders * 100 #calculating the cancellation rate
print("Cancellation Rate:",round(CancellationRate, 2))
ReturnRate = Df[Df['Status'] == 'Returned'].shape[0] / AllOrders * 100 #calculating the return rate
print("Return Rate:", round(ReturnRate, 2))
FailRate = Df[Df['PaymentStatus'] == 'Failed'].shape[0] / AllOrders * 100 #calculating the failure rate
print("Failure Rate:", round(FailRate, 2))
```
